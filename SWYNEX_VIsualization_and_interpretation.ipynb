```python
from Bio import SeqIO

file_name = next(iter(uploaded))

record = SeqIO.read(file_name, "fasta")

sequence = str(record.seq).upper()

print("Sequence ID:", record.id)
print("DNA Sequence Length:", len(sequence))
print("DNA Sequence:")
print(sequence)
```

```python
from collections import Counter

# Count DNA bases
base_counts = Counter(sequence)

print("DNA BASE COMPOSITION")
print("--------------------")

print("Adenine (A):", base_counts["A"])
print("Thymine (T):", base_counts["T"])
print("Guanine (G):", base_counts["G"])
print("Cytosine (C):", base_counts["C"])

# Calculate GC and AT content
valid_bases = sum(base_counts[b] for b in ["A", "T", "G", "C"])

if valid_bases > 0:
    gc_content = (
        (base_counts["G"] + base_counts["C"])
        / valid_bases
    ) * 100

    at_content = (
        (base_counts["A"] + base_counts["T"])
        / valid_bases
    ) * 100

    print("\nGC Content:", round(gc_content, 2), "%")
    print("AT Content:", round(at_content, 2), "%")
else:
    print("No valid DNA bases found.")
```

```python
import matplotlib.pyplot as plt

# DNA bases and their counts
bases = ["A", "T", "G", "C"]
counts = [base_counts[base] for base in bases]

# Create the bar chart
plt.figure(figsize=(8, 5))
plt.bar(bases, counts)

plt.title("DNA Base Composition")
plt.xlabel("DNA Bases")
plt.ylabel("Number of Bases")

# Save the chart
plt.savefig("DNA_Base_Composition.png", dpi=300)

# Display the chart
plt.show()

print("DNA Base Composition chart created successfully!")
```

```python
import matplotlib.pyplot as plt

# Create the GC vs AT content graph
plt.figure(figsize=(8, 5))

plt.bar(
    ["GC Content", "AT Content"],
    [gc_content, at_content]
)

plt.title("GC Content vs AT Content")
plt.xlabel("Base Composition")
plt.ylabel("Percentage (%)")
plt.ylim(0, 100)

# Save the graph
plt.savefig("GC_AT_Content.png", dpi=300)

# Display the graph
plt.show()

print("GC Content:", round(gc_content, 2), "%")
print("AT Content:", round(at_content, 2), "%")
print("Second graph created successfully!")
```

```python
import matplotlib.pyplot as plt

# Calculate DNA sequence length
sequence_length = len(sequence)

# Create the graph
plt.figure(figsize=(7, 5))

plt.bar(
    ["DNA Sequence"],
    [sequence_length]
)

plt.title("DNA Sequence Length")
plt.xlabel("Sequence")
plt.ylabel("Number of Bases")

# Save the graph
plt.savefig("DNA_Sequence_Length.png", dpi=300)

# Display the graph
plt.show()

print("DNA Sequence Length:", sequence_length)
print("Third graph created successfully!")
```
